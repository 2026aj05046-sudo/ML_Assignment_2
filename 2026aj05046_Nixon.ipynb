```python
                                            Training and Testing data
```

```python

from sklearn.model_selection import train_test_split

# Select 30 input features.
X = df.drop(columns=["diagnosis"])

# Encode target:
# B = Benign (0), M = Malignant (1)
y = df["diagnosis"].map({"B": 0, "M": 1})

assert y.notna().all(), "Unexpected diagnosis labels"

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
print("Number of features:", X_train.shape[1])
```

```python
Importing required binaries
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

print("All five classification models imported successfully!")
```

```python
defining the all five models
```

```python

models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=2000,
            random_state=42
        ))
    ]),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "KNN": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", KNeighborsClassifier(
            n_neighbors=5
        ))
    ]),

    "Naive Bayes": GaussianNB(),

    "Support Vector Machine": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", SVC(
            kernel="rbf",
            probability=True,
            random_state=42
        ))
    ])
}

print("Total models configured:", len(models))
```

```python
                                        Training all the 5 models
```

```python

for name, model in models.items():

    print("Training:", name)

    model.fit(X_train, y_train)

    print("Training completed successfully!\n")

print("ALL FIVE MODELS TRAINED SUCCESSFULLY!")
```

```python
Metrics calculations
```

```python

import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

results = []

for name, model in models.items():

    # Predict classes using test data
    y_pred = model.predict(X_test)

    # Probability of malignant class (1)
    positive_index = list(model.classes_).index(1)
    y_prob = model.predict_proba(X_test)[:, positive_index]

    # Calculate six evaluation metrics
    result = {
        "ML Model Name": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "AUC": roc_auc_score(y_test, y_prob),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1": f1_score(
            y_test, y_pred, zero_division=0
        ),
        "MCC": matthews_corrcoef(y_test, y_pred)
    }

    results.append(result)

    print("\nModel:", name)
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))

    print("\nClassification Report:")
    print(classification_report(
        y_test,
        y_pred,
        target_names=["Benign", "Malignant"],
        zero_division=0
    ))

comparison_df = pd.DataFrame(results)

print("\nFINAL MODEL COMPARISON")
display(comparison_df.round(4))
```

```python

# ==========================================
# 2026aj05046-NIXON.M - ML ASSIGNMENT 2
# Load, Train and Evaluate 5 ML Models
# ==========================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

import joblib
import os

# STEP 1: Load dataset
df = pd.read_csv("breast_cancer_dataset.csv")

print("Dataset loaded:", df.shape)

# STEP 2: Separate input and target
X = df.drop(columns=["diagnosis"])
y = df["diagnosis"].map({"B": 0, "M": 1})

assert y.notna().all(), "Unexpected target labels"

# STEP 3: Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

# STEP 4: Define five models
models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=2000, random_state=42
        ))
    ]),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "KNN": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", KNeighborsClassifier(
            n_neighbors=5
        ))
    ]),

    "Naive Bayes": GaussianNB(),

    "Support Vector Machine": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", SVC(
            kernel="rbf",
            probability=True,
            random_state=42
        ))
    ])
}

# STEP 5: Train and evaluate
results = []

for name, model in models.items():

    print("\nTraining:", name)

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    positive_index = list(model.classes_).index(1)
    y_prob = model.predict_proba(X_test)[:, positive_index]

    results.append({
        "ML Model Name": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "AUC": roc_auc_score(y_test, y_prob),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1": f1_score(
            y_test, y_pred, zero_division=0
        ),
        "MCC": matthews_corrcoef(y_test, y_pred)
    })

    print("Training and evaluation completed!")

    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))

    print("Classification Report:")
    print(classification_report(
        y_test, y_pred,
        target_names=["Benign", "Malignant"],
        zero_division=0
    ))

# STEP 6: Create comparison table
comparison_df = pd.DataFrame(results)

print("\nFINAL MODEL COMPARISON")
display(comparison_df.round(4))

# STEP 7: Create output folders
os.makedirs("model", exist_ok=True)
os.makedirs("results", exist_ok=True)

# STEP 8: Save all five models
model_files = {
    "Logistic Regression": "logistic_regression.joblib",
    "Decision Tree": "decision_tree.joblib",
    "KNN": "knn.joblib",
    "Naive Bayes": "naive_bayes.joblib",
    "Support Vector Machine": "svm.joblib"
}

for name, filename in model_files.items():
    joblib.dump(models[name], os.path.join("model", filename))
    print("Saved:", filename)

# STEP 9: Save comparison results
comparison_df.to_csv(
    "results/model_comparison.csv",
    index=False
)

# STEP 10: Save labeled test data
test_data = X_test.copy()
test_data["diagnosis"] = y_test

test_data.to_csv("test_data.csv", index=False)

print("\nTest data saved:", test_data.shape)

# STEP 11: Verify saved models without retraining
for name, filename in model_files.items():
    loaded_model = joblib.load(os.path.join("model", filename))
    loaded_predictions = loaded_model.predict(X_test)

    original_predictions = models[name].predict(X_test)

    assert np.array_equal(
        loaded_predictions, original_predictions
    )

    print(name, "- Saved model verified")

# STEP 12: Identify winner
ranked = comparison_df.sort_values(
    by=["MCC", "F1", "AUC"],
    ascending=False
)

print("\nNixonWINNER:")
print(ranked.iloc[0]["ML Model Name"])

print("\nASSIGNMENT with the STEP 2 COMPLETED SUCCESSFULLY! by Nixon M")
```
